```python
# college_regression_validation.py
# Complete script to reproduce the workflow you requested.
# Requirements: pandas, scikit-learn, openpyxl (for reading .xlsx)
# Install with: pip install pandas scikit-learn openpyxl

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

def main():
    # ---- CONFIGURE FILE PATH ----
    # Update this path if your Excel file is in a different location.
    path = "/Users/sudhirsingh/Downloads/f4aa1250-3c0a-480b-bd6b-c953fa0e987c-College.xlsx"

    # ---- LOAD DATA ----
    df = pd.read_excel(path)

    # ---- PREPROCESSING ----
    # 1) Remove the first column (university name)
    df = df.drop(df.columns[0], axis=1)

    # 2) One-hot encode 'Private', drop_first=True
    df = pd.get_dummies(df, columns=['Private'], drop_first=True)

    # ---- DESCRIPTIVE STATISTICS ----
    desc = df.describe().round(2)
    print("Descriptive statistics (rounded to 2 decimals):\n")
    print(desc)
    print("\n" + "="*80 + "\n")

    # ---- PREPARE FEATURES & TARGET ----
    target = 'Grad.Rate'
    X = df.drop(columns=[target])
    y = df[target]

    # ---- SPLIT (70:30) with random_state=42 ----
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.30, random_state=42
    )

    # ---- FIT LINEAR REGRESSION ----
    model = LinearRegression()
    model.fit(X_train, y_train)

    # ---- MODEL COEFFICIENTS (rounded to 2 decimals using pandas .round()) ----
    coef_series = pd.Series(model.coef_, index=X.columns).sort_index()
    coef_df = coef_series.to_frame(name='coefficient').round(2)
    print("Model coefficients (rounded to 2 decimals):\n")
    print(coef_df)
    print("\n" + "="*80 + "\n")

    # ---- PREDICTIONS ----
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # ---- METRICS (rounded to 2 decimals using round()) ----
    r2_train = round(r2_score(y_train, y_train_pred), 2)
    mse_train = round(mean_squared_error(y_train, y_train_pred), 2)
    mae_train = round(mean_absolute_error(y_train, y_train_pred), 2)

    r2_test = round(r2_score(y_test, y_test_pred), 2)
    mse_test = round(mean_squared_error(y_test, y_test_pred), 2)
    mae_test = round(mean_absolute_error(y_test, y_test_pred), 2)

    print("Performance metrics (rounded to 2 decimals):\n")
    print(f"Train R² : {r2_train}")
    print(f"Train MSE: {mse_train}")
    print(f"Train MAE: {mae_train}\n")
    print(f"Test R²  : {r2_test}")
    print(f"Test MSE : {mse_test}")
    print(f"Test MAE : {mae_test}")

    # ---- OPTIONAL: Save outputs to CSV for inspection ----
    coef_df.to_csv("linear_model_coefficients.csv", index=True)
    desc.to_csv("descriptive_statistics.csv")
    preds_train = pd.DataFrame({'y_train': y_train, 'y_train_pred': y_train_pred})
    preds_test  = pd.DataFrame({'y_test': y_test, 'y_test_pred': y_test_pred})
    preds_train.to_csv("train_predictions.csv", index=False)
    preds_test.to_csv("test_predictions.csv", index=False)
    print("\nSaved: linear_model_coefficients.csv, descriptive_statistics.csv, train_predictions.csv, test_predictions.csv")

if __name__ == "__main__":
    main()
```
